```python
from pathlib import Path

import pandas as pd
```

```python
DATA_DIR = Path.cwd().resolve().parent / 'data'
data = pd.read_csv(DATA_DIR / 'sequence_craft.csv')
```

```python
data = data[['name', 'e', 'kobs']].rename(columns={'e': 'sequence', 'kobs': 'target'})
data['sequence'] = data['sequence'].astype('string').str.upper().str.strip()
data['target'] = pd.to_numeric(data['target'], errors='coerce')
data = data.dropna(subset=['sequence', 'target'])
data = data[data['sequence'].str.fullmatch(r'[ACGT]+', na=False)]
data = data.drop_duplicates(subset='sequence').reset_index(drop=True)
data['target_binary'] = (data['target'] > 1e-7).astype(int)
data
```

```python
data.to_csv(DATA_DIR / 'sequence_craft_clean.csv', index=False)
```
